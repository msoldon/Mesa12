### Perfil dos regimes

A tabela abaixo mostra as **medianas nas unidades originais**. Para facilitar a leitura, os regimes são ordenados de menor para maior intensidade conjunta dos principais poluentes/sensores.

```python
profile_rows = []
for r in sorted(np.unique(regimes)):
    idx = regimes == r
    row = {
        'Regime': int(r),
        'N': int(idx.sum()),
        '%': 100 * float(idx.mean()),
    }
    for j, name in enumerate(feature_names):
        row[name] = float(np.median(X_imputed[idx, j]))
    profile_rows.append(row)

cols = ['Regime', 'N', '%'] + list(feature_names)
html = ['<div style="overflow-x:auto"><table><tr>']
html += [f'<th>{c}</th>' for c in cols]
html.append('</tr>')
for row in profile_rows:
    html.append('<tr>')
    for c in cols:
        v = row[c]
        if c == 'Regime' or c == 'N':
            txt = f'{int(v):,}'
        elif c == '%':
            txt = f'{v:.1f}%'
        else:
            txt = f'{v:.2f}'
        html.append(f'<td>{txt}</td>')
    html.append('</tr>')
html.append('</table></div>')
display(HTML(''.join(html)))
```

```python
runs = []
start = 0
for i in range(1, len(regimes) + 1):
    if i == len(regimes) or regimes[i] != regimes[start]:
        runs.append((int(regimes[start]), start, i - 1, i - start))
        start = i

html = ['<table><tr><th>Regime</th><th>Episódios</th><th>Duração mediana (h)</th><th>Duração média (h)</th><th>Máxima (h)</th></tr>']
for r in sorted(np.unique(regimes)):
    lengths = np.array([x[3] for x in runs if x[0] == r])
    html.append(
        f'<tr><td>{r}</td><td>{len(lengths)}</td>'
        f'<td>{np.median(lengths):.1f}</td><td>{np.mean(lengths):.1f}</td><td>{np.max(lengths)}</td></tr>'
    )
html.append('</table>')
display(HTML(''.join(html)))
print('Transições de regime:', len(runs) - 1)
```
