# Descoberta de regimes operacionais e anomalias com CRISP-DM

## 1. Entendimento do negócio

**Pergunta de negócio**
> *“Nossa operação tem regimes distintos que ninguém mapeou, e comportamentos atípicos que só percebemos depois da falha. Que regimes existem, e o que é genuinamente anômalo?”*

### Objetivos analíticos
- descobrir quantos **macro-regimes** estão realmente suportados pelos dados;
- caracterizar cada regime pelas variáveis que mais o distinguem;
- identificar **transições de regime** ao longo do tempo;
- separar **problemas de qualidade de dados** de **anomalias comportamentais**;
- detectar anomalias **condicionadas ao regime**, reduzindo falsos positivos.

### Critérios de sucesso
Um resultado será considerado útil se:
- os clusters tiverem separação mensurável (Silhouette, Davies–Bouldin e Calinski–Harabasz);
- nenhum regime escolhido for apenas um microgrupo residual sem representatividade;
- os regimes puderem ser descritos em linguagem operacional;
- as anomalias forem raras, rastreáveis no tempo e não decorram simplesmente de valores ausentes.

### Princípio central
Uma observação **não deve ser chamada de anômala apenas por pertencer a um regime raro**. Primeiro descobrimos os estados recorrentes; depois avaliamos o quão estranho cada ponto é **comparado com seu próprio regime**.
