# Documentação do Dataset: Air Quality UCI

## 1. Visão Geral do Dataset
O conjunto de dados **AirQualityUCI** contém 9.358 observações horárias de respostas de um conjunto de 5 sensores químicos de óxido metálico incorporados em um dispositivo multissensor de qualidade do ar.
O dispositivo foi instalado em campo em uma área altamente poluída, ao nível da rua, em uma cidade italiana. As gravações foram realizadas de **março de 2004 a abril de 2005** (1 ano).

## 3. Peculiaridades e Regras de Qualidade

### Codificação de Valores Ausentes
- **`-200`**: No dataset bruto, valores ausentes ou falhas de leitura do sensor estão codificados estritamente como **`-200`**.
- **Ação obrigatória**: Todos os valores iguais a `-200` devem ser convertidos para `NaN` antes de qualquer cálculo de média, desvio padrão ou modelagem.

### Regras de Qualidade CRISP-DM Adotadas:
1. **Descarte de Colunas (>50% de Nulos)**:
   - A variável `NMHC(GT)` possui mais de 90% de valores ausentes (`-200`). Imputá-la dominaria a informação real. Portanto, é removida do pipeline de modelagem.
2. **Filtro de Baixa Qualidade de Registro (>30% nulos na linha)**:
   - Linhas com mais de 30% das variáveis ausentes são marcadas como **Baixa Qualidade de Dado**.
   - **Princípio Importante**: Falha de sensor/leitura ausente **NÃO É ANOMALIA OPERACIONAL**. Registros de baixa qualidade são isolados para não gerar alertas falsos de anomalia.
3. **Estratégia de Imputação**:
   - Interpolação linear curta para lacunas continuas de até **3 horas** (captura dinâmica temporal recente).
   - **KNN Imputer (k=5)** para lacunas remanescentes maiores.

## 4. Contexto de Negócio & Aprendizado Não Supervisionado

- **Regimes Operacionais (Clusters)**: Descobrem os estados padrão de poluição e clima (ex: Horário de Pico, Tráfego Moderado, Madrugada/Baixa Atividade).
- **Detecção de Anomalias Condicionada ao Regime**: Treina um modelo *Isolation Forest* independente para cada regime. Uma medição só é considerada anômala se for extrema **em relação ao seu próprio regime**, eliminando falsos positivos decorrentes de mudanças normais de estado.
